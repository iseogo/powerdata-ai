```python
# Import Power Data AI components
import sys
sys.path.append('..')  # Add parent directory to path

from powerdata.core.analyzer import DataAnalyzer
from powerdata.visualizations.charts import ChartBuilder
from powerdata.core.reporter import ReportGenerator

import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

print("✅ Power Data AI components loaded successfully!")
```

## 1. Load the Data

```python
# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]

print(f"📊 Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns")
df.head()
```

## 2. Data Analysis with Power Data AI

```python
# Initialize analyzer
analyzer = DataAnalyzer(df)

# Get overview
overview = analyzer.get_overview()

print("📋 DATASET OVERVIEW")
print("=" * 60)
print(f"Shape: {overview['shape']}")
print(f"Columns: {overview['columns']}")
print(f"Duplicates: {overview['duplicates']}")
print(f"Memory: {overview['memory_usage']}")
print("=" * 60)
```

```python
# Get key insights
insights = analyzer.get_key_insights()

print("💡 KEY INSIGHTS")
print("=" * 60)
for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")
print("=" * 60)
```

```python
# Statistical summary
stats = analyzer.get_statistics()
print("📈 STATISTICAL SUMMARY")
print("=" * 60)
stats
```

```python
# Find correlations
correlations = analyzer.find_correlations(threshold=0.7)

print("🔗 STRONG CORRELATIONS")
print("=" * 60)
for corr in correlations:
    print(f"{corr['column1']:30} <-> {corr['column2']:30} : {corr['correlation']:6.3f} ({corr['strength']})")
print("=" * 60)
```

## 3. Visualizations with Brand Styling

```python
# Initialize chart builder
builder = ChartBuilder()

# Distribution of petal length
fig1 = builder.histogram(df, 'petal length (cm)', title='Distribution of Petal Length')
fig1.show()
```

```python
# Species distribution
species_counts = df['species'].value_counts()
fig2 = builder.bar_chart(species_counts, title='Species Distribution')
fig2.show()
```

```python
# Petal length vs width by species
fig3 = builder.scatter_plot(
    df, 
    'petal length (cm)', 
    'petal width (cm)', 
    color='species',
    title='Petal Dimensions by Species'
)
fig3.show()
```

```python
# Correlation heatmap
fig4 = builder.correlation_heatmap(df, title='Feature Correlations')
fig4.show()
```

```python
# Box plot comparison
fig5 = builder.box_plot(
    df, 
    'petal length (cm)', 
    group_by='species',
    title='Petal Length Distribution by Species'
)
fig5.show()
```

## 4. Generate Professional Report

```python
# Initialize report generator
report = ReportGenerator(df, "Iris Dataset Analysis - Power Data AI")

# Add executive summary
report.add_executive_summary(
    "This analysis examines the classic Iris dataset containing measurements of 150 iris flowers "
    "across three species (Setosa, Versicolor, and Virginica). The dataset demonstrates clear "
    "patterns in petal and sepal dimensions that enable accurate species classification. "
    "Our analysis reveals strong correlations between petal features and identifies Setosa "
    "as clearly distinguishable from the other two species based on morphological characteristics."
)

# Add insights
report.add_insights([
    "🌸 Petal length and width show the strongest discriminative power between species",
    "📊 Setosa is clearly separable with smaller petal dimensions (avg 1.5cm length vs 4.3cm for others)",
    "🔗 Extremely high correlation (0.96) between petal length and width across all species",
    "✅ Dataset quality is excellent: no missing values, perfectly balanced classes (50 samples each)",
    "📈 Petal measurements alone can achieve near-perfect classification accuracy"
])

# Add statistics
report.add_statistics(stats)

# Add recommendations
report.add_recommendation([
    "Use petal length and width as primary features for any species classification model",
    "Start with simple models (Logistic Regression, Decision Trees) - the data is linearly separable",
    "Focus additional data collection on Versicolor and Virginica to improve inter-species separation",
    "Consider this dataset as a benchmark for testing new classification algorithms",
    "Apply similar analysis methodology to other botanical classification problems"
])

print("✅ Report sections added successfully!")
```

```python
# Generate and save HTML report
report.save_html('../examples/reports/iris_analysis_report.html')
print("✅ HTML report saved to: examples/reports/iris_analysis_report.html")

# Generate and save Markdown report
report.save_markdown('../examples/reports/iris_analysis_report.md')
print("✅ Markdown report saved to: examples/reports/iris_analysis_report.md")
```
